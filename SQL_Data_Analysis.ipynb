####Data exploration

```python
##Write a SQL query to retrieve all columns for sales made on '2022-11-05
run("""
SELECT *
FROM retail_sales
WHERE sale_date = '2022-11-05';
""" , retail)
```

```python
##Write a SQL query to retrieve all transactions where the category is 'Beauty' and the quantity sold is more than 4 in the duration of '2022-11-01'  '2022-11-30'

run("""
SELECT*
FROM retail_sales
WHERE category ='Beauty'
AND quantity >=4
AND sale_date BETWEEN '2022-11-01' AND '2022-11-30'
""" , retail)
```

```python
##Write a SQL query to retrieve all transactions where the category is 'Clothing' and the quantity sold is more than 4 in the month of Nov-2022
run("""
SELECT *
FROM retail_sales
WHERE category = 'Clothing'
  AND strftime('%Y-%m', sale_date) = '2022-11'
  AND quantity >= 4
""", retail)
```

```python
##Write a SQL query to calculate the total sales  for each category.
run("""SELECT
    category,
    SUM(total_sale) as net_sale,
    COUNT(*) as total_orders
FROM retail_sales
GROUP BY 1 """, retail)
```

```python
##Write a SQL query to find the average age of customers who purchased items from the 'Beauty' category.
run("""
SELECT ROUND(AVG(age), 2) AS avg_age
FROM retail_sales
WHERE category = 'Beauty'
""", retail)
```

```python
##Write a SQL query to find all transactions where the total_sale is greater than 1000.
run("""
SELECT * FROM retail_sales
WHERE total_sale >1000

""" ,retail)
```

```python
## Write a SQL query to find the total number of transactions made by each gender in each category
run("""
SELECT
category,
gender,
 COUNT(*) as total_trans
FROM retail_sales
GROUP
    BY
    category,
    gender
ORDER BY 1

""", retail)
```

```python
##Write a SQL query to calculate the average sale for each month. Find out best selling month in each year
run("""
SELECT year, month, avg_sale
FROM (
    SELECT
        strftime('%Y', sale_date) AS year,
        strftime('%m', sale_date) AS month,
        ROUND(AVG(total_sale), 2) AS avg_sale,
        RANK() OVER (
            PARTITION BY strftime('%Y', sale_date)
            ORDER BY AVG(total_sale) DESC
        ) AS rnk
    FROM retail_sales
    GROUP BY year, month
) AS t
WHERE rnk = 1
""", retail)
```

```python
##Write a SQL query to find the top 5 customers based on the highest total sales
run("""
SELECT
    customer_id,
    SUM(total_sale) AS total_sales
FROM retail_sales
GROUP BY customer_id
ORDER BY total_sales DESC
LIMIT 5
""" , retail)
```

```python
##Write a SQL query to find the number of unique customers who purchased items from each category.
run("""
SELECT
category,
COUNT(DISTINCT customer_id ) as unique_cs
FROM retail_sales
GROUP BY category
""" ,retail)
```

```python
##Write a SQL query to create each shift and number of orders (Example Morning <12, Afternoon Between 12 & 17, Evening >17)
run("""
SELECT
    CASE
        WHEN CAST(strftime('%H', sale_time) AS INTEGER) < 12 THEN 'Morning'
        WHEN CAST(strftime('%H', sale_time) AS INTEGER) BETWEEN 12 AND 17 THEN 'Afternoon'
        ELSE 'Evening'
    END AS shift,
    COUNT(*) AS num_orders
FROM retail_sales
GROUP BY shift
""", retail)
```
