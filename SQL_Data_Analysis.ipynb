```python
from google.colab import drive
import sqlite3, pandas as pd, shutil, os

drive.mount('/content/drive')

DRIVE = '/content/drive/MyDrive/'
LOCAL = '/content'


def run(query, conn):

    return pd.read_sql(query, conn)
```

```python
df = pd.read_csv(f'{DRIVE}/Retail_Sales_Analytics.csv')
```

```python
df = df.rename(columns={'quantiy': 'quantity'})

retail = sqlite3.connect(f'{LOCAL}/retail.db')
df.to_sql('retail_sales', retail, if_exists='replace', index=False)

print(run("SELECT COUNT(*) AS rows_loaded FROM retail_sales", retail))
run("SELECT * FROM retail_sales LIMIT 5", retail)
```

```python
run("""
SELECT * FROM retail_sales
WHERE transactions_id is NULL

""" ,retail)
```

```python
run("""
SELECT * FROM retail_sales
WHERE age IS NULL
   OR quantity IS NULL
   OR price_per_unit IS NULL
   OR cogs IS NULL
   OR total_sale IS NULL
""", retail)
```

####Data exploration

```python
##Write a SQL query to retrieve all columns for sales made on '2022-11-05
run("""
SELECT *
FROM retail_sales
WHERE sale_date = '2022-11-05';
""" , retail)
```

```python
##Write a SQL query to retrieve all transactions where the category is 'Beauty' and the quantity sold is more than 4 in the duration of '2022-11-01'  '2022-11-30'

run("""
SELECT*
FROM retail_sales
WHERE category ='Beauty'
AND quantity >=4
AND sale_date BETWEEN '2022-11-01' AND '2022-11-30'
""" , retail)
```

```python
##Write a SQL query to retrieve all transactions where the category is 'Clothing' and the quantity sold is more than 4 in the month of Nov-2022
run("""
SELECT *
FROM retail_sales
WHERE category = 'Clothing'
  AND strftime('%Y-%m', sale_date) = '2022-11'
  AND quantity >= 4
""", retail)
```

```python
##Write a SQL query to calculate the total sales  for each category.
run("""SELECT
    category,
    SUM(total_sale) as net_sale,
    COUNT(*) as total_orders
FROM retail_sales
GROUP BY 1 """, retail)
```

```python
##Write a SQL query to find the average age of customers who purchased items from the 'Beauty' category.
run("""
SELECT ROUND(AVG(age), 2) AS avg_age
FROM retail_sales
WHERE category = 'Beauty'
""", retail)
```

```python
##Write a SQL query to find all transactions where the total_sale is greater than 1000.
run("""
SELECT * FROM retail_sales
WHERE total_sale >1000

""" ,retail)
```

```python
## Write a SQL query to find the total number of transactions made by each gender in each category
run("""
SELECT
category,
gender,
 COUNT(*) as total_trans
FROM retail_sales
GROUP
    BY
    category,
    gender
ORDER BY 1

""", retail)
```
